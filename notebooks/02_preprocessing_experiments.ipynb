```python
# Import path handling, pandas, and the project's configurable text cleaner.
from pathlib import Path

import pandas as pd

from support_sense.preprocessing import preprocess_text
```

```python
# Point to the raw dataset using the path expected when running from notebooks/.
DATA_PATH = Path("../data/support_tickets.csv")
```

```python
# Load the support-ticket records into a DataFrame.
df = pd.read_csv(DATA_PATH)
```

```python
# Create a cleaned text feature while keeping the original message unchanged.
df["processed_text"] = df["text"].apply(
    lambda text: preprocess_text(text, lowercase=True, strip_punctuation=True)
)
```

```python
# Compare the raw and processed text in the first five records.
df.head()
```

```python
# Define one example containing extra whitespace, mixed case, punctuation, and a URL.
raw = "  My PAYMENT failed!!! Visit https://example.com/help  "
```

```python
# Show the conservative defaults: lowercase and normalize whitespace only.
print(preprocess_text(raw))
```

```python
# Compare a stronger configuration that also removes URLs and punctuation.
print(
    preprocess_text(
        raw,
        strip_urls=True,
        strip_punctuation=True,
    )
)
```
